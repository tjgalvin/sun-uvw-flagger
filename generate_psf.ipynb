```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from casacore.tables import table, taql
from jolly_roger.tractor import get_baseline_data, get_open_ms_tables
from jolly_roger.uvws import get_object_delay_for_ms
import fixms

from astropy.visualization import (
        ImageNormalize,
        SqrtStretch,
        LogStretch,
        ZScaleInterval,
        quantity_support,
        time_support,
        MinMaxInterval,
    )
```

```python
import fixms.fix_ms_dir


ms_path = Path("scienceData.EMU_1141-55.SB47138.EMU_1141-55.beam00_averaged_cal.leakage.ms_trans")

fixms.fix_ms_dir.fix_ms_dir(ms=str(ms_path))

open_ms_tables = get_open_ms_tables(ms_path=ms_path)
baseline_data = get_baseline_data(open_ms_tables=open_ms_tables, ant_1=0, ant_2=1)


delays = get_object_delay_for_ms(ms_path=ms_path)

print(f"{delays=}")

```

```python

# baseline_data.masked_data.mask = np.zeros(
#     baseline_data.masked_data.shape
# )

print(f"{delays.w_delays.shape=}")

with quantity_support(), time_support():
    data = np.abs(baseline_data.masked_data[..., 0])
    norm = ImageNormalize(
        data, interval=ZScaleInterval(), stretch=SqrtStretch()
    )
    cmap = plt.cm.viridis

    fig, ax = plt.subplots(1,1)

    extract = 2500
    print(f"{delays.w_delays.shape=}")
    baseline_id = delays.b_map[(0, 1)]
    print(f"{baseline_id=}")
    print(f"{baseline_data.freq_chan[:10]=}")
    expected_delays = delays.w_delays[baseline_id, extract]

    ax.pcolormesh(
        np.arange(baseline_data.time.shape[0]),
        baseline_data.freq_chan,
        data.T,
        norm=norm,
        cmap=cmap
    )
    ax.axvline(extract, color="red")
    
    freqs = baseline_data.freq_chan
    time_data = baseline_data.masked_data[extract]
    i = time_data[:, 0] + time_data[:, 3]
    
    SAMPLE = 100000
    
    delay = np.fft.fftshift(
        np.fft.ifft(i.filled(0+0j), n=SAMPLE)
    )
    time = np.fft.fftshift(
        np.fft.fftfreq(SAMPLE, d=np.mean(-np.diff(freqs)))
    ).decompose()
    
    bandpass = 1+0j - time_data.mask

    psf = np.fft.fftshift(
        np.fft.ifft(bandpass[:, 0] , n=SAMPLE), 
    )
    psf_time = np.fft.fftshift(
        np.fft.fftfreq(len(psf), d=np.mean(-np.diff(freqs)))
    ).decompose()
    
    print(f"{time_data.shape=}")
    
    fig, axes = plt.subplots(2,1)
    ax1, ax2 = axes.flatten()
    
    ax1.plot(
        freqs,
        np.abs(time_data[:, 0])
    )
    ax2.plot(
        time,
        np.abs(delay)
    )
    ax2.axvline(expected_delays, color="red")
    print(expected_delays)
    
    max_delay = np.max(np.abs(delay))
    
    _shift = np.argmin(np.abs(time - expected_delays)) - np.argmin(np.abs(time - 0)) 
    psf = psf / np.max(psf)
    ax2.plot(
        time,
        np.abs(
            np.roll(psf, _shift) * max_delay
        )
    )
    print(f"{psf.shape=}")
    print(f"{_shift=}")
    ax2.set(
        xlim=[-2e-7, 0]
    )
    
```

```python
plt.close("all")

above_idx = 2200
TIMESTEPS = 1000
SAMPLES = 100000
MID = SAMPLES // 2
WINDOW = int(MID*0.1)

print(baseline_data.freq_chan.shape)

subset_time = np.fft.fftshift(
    np.fft.fftfreq(n=SAMPLES, d=np.mean(np.diff(baseline_data.freq_chan)))
).decompose()

with quantity_support(), time_support():


    fig, ax = plt.subplots(1,1)

    data = []
    for c, idx in enumerate(range(above_idx, above_idx + TIMESTEPS)):
        subset_data = baseline_data.masked_data[idx]

        subset_i = (subset_data[..., 0] + subset_data[..., 1]) / 2
        subset_delay = np.fft.fftshift(
            np.fft.ifft(subset_i.filled(0+0j), n=SAMPLES)
        )
        shift = np.argmax(np.abs(subset_delay))
        # subset_delay.real = np.roll(subset_delay.real, -shift + MID)
        # subset_delay.imag = np.roll(subset_delay.imag, -shift + MID)
        subset_delay = np.roll(subset_delay, -shift + MID)
        
        subset_delay = subset_delay / np.max(np.abs(subset_delay))
        
        ax.plot(
            subset_time,
            np.abs(subset_delay),
            color="black",
            alpha=0.4,
            label="Data" if c == 0 else None
        )

        data.append(subset_delay)


    median = np.median(
        np.abs(data), axis=0
    )

    ax.plot(subset_time, median, color="red", label="Median")
    ax.legend()
    
    ax.set(
        xlim=[subset_time[MID-WINDOW], subset_time[MID+WINDOW]],
        title=f"Stacked Data ({TIMESTEPS} timesteps)",
        ylabel="Normalised Units",

    )
    ax.grid()


median_psf = median

```
